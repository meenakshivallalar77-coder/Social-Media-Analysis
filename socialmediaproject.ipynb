```python
#1. See the first 10 rows of a table
query = "SELECT * FROM user_details LIMIT 10;"
df1 = pd.read_sql(query, conn)
df1
```

```python
#2. Count total rows in a table
query = "SELECT COUNT(*) AS total_users FROM user_details;"
df2 = pd.read_sql(query, conn)
df2
```

```python
#3. Find users from a specific country
query = "SELECT full_name, country, platform_primary FROM user_details WHERE country = 'USA';"
df3 = pd.read_sql(query, conn)
df3
```

```python
#4. List influencers sorted by followers (highest first)
query = "SELECT influencer_name, platform, followers_count FROM influencer_details_clean ORDER BY followers_count DESC LIMIT 10;"
df4 = pd.read_sql(query, conn)
df4
```

```python
#5. Find all 1-star reviews
query = "SELECT review_id, platform, review_text FROM customer_reviews WHERE rating = 1;"
df5 = pd.read_sql(query, conn)
df5
```

```python
#6. Total ad revenue by platform
query = "SELECT platform, SUM(revenue_usd) AS total_revenue FROM ad_revenue_details GROUP BY platform ORDER BY total_revenue DESC;"
df6 = pd.read_sql(query, conn)
df6
```

```python
#7. Count how many posts are flagged as spam vs not spam
query = "SELECT spam_flag, COUNT(*) AS post_count FROM content_quality GROUP BY spam_flag;"
df7 = pd.read_sql(query, conn)
df7
```

```python
#8. Count open vs resolved support tickets
query = "SELECT status, COUNT(*) AS ticket_count FROM support_tickets_details GROUP BY status;"
df8 = pd.read_sql(query, conn)
df8
```

```python
#9. Find top 5 trending hashtags by usage count
query = "SELECT hashtag, platform, usage_count FROM hashtag_trend_details ORDER BY usage_count DESC LIMIT 5;"
df9 = pd.read_sql(query, conn)
df9
```

```python
#10. Find the top 5 highest-earning ad transactions
query = "SELECT platform, ad_type, revenue_usd, region FROM ad_revenue_details ORDER BY revenue_usd DESC LIMIT 5;"
df10 = pd.read_sql(query, conn)
df10
```

```python
query = "SELECT platform, AVG(rating) AS avg_rating FROM customer_reviews GROUP BY platform ORDER BY avg_rating DESC;"
df_platform_rating = pd.read_sql(query, conn)
df_platform_rating
```

```python
import matplotlib.pyplot as plt

plt.bar(df_platform_rating['platform'], df_platform_rating['avg_rating'], color='brown')
plt.xlabel('Platform')
plt.ylabel('Average Rating')
plt.title('Average Customer Rating by Platform')
plt.xticks(rotation=45)
plt.ylim(0, 5)
plt.show()
```
